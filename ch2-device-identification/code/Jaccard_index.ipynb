```python
#apply jaccard_index rule
from textdistance import levenshtein
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
import time
```

```python
def fuzzy_intersection(a, b):
    if len(a) > len(b):
        a, b = b, a
    s = 0
    for i in range(len(a)):
        print(str(a[i]), str(b[i]))
        print("lev=",levenshtein(str(a[i]), str(b[i])))
        print("max=",max(len(str(a[i])), len(str(b[i]))))
        m = levenshtein.normalized_similarity(str(a[i]), str(b[i]))
        print("m=",m)
        if m:
            s += m
            print("s=",s)
    return s

```

```python
def jaccard_index(a, b):
    intersection = fuzzy_intersection(a, b)
    print("intersection=",intersection)
    union = len(a) + len(b) - intersection
    print("union=",union)
    jaccard_index = intersection / union
    return jaccard_index
```

```python
def find_matching_device_types(b, database):
    result = []
    for db in range(len(database)):
        device_type = database.iloc[db].type
        a = database.iloc[db].drop(['type'])
        j_index = jaccard_index(a, b)
        result.append((j_index, device_type))
    return sorted(result, key=lambda x: x[0], reverse=True)

```
